# Exercise 2

```python
number1 = int(input("Write the first number: "))
number2 = int(input("Write the second number: "))
add = number1 + number2
rest = number1 - number2
multiplication = number1 * number2
divition = number1 / number2
print("The sum is: " + str(add))
print("The difference is: " + str(rest))
print("The product is: " + str(multiplication))
print("The quotient is: " + str(divition)) 
```
